# Back-translate

```python
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
from pydna.dseq import Dseq

# Simplified back-translation codon table
back_translation_code = {
        'A': ['GCA', 'GCC', 'GCG', 'GCT'],
        'C': ['TGT', 'TGC'],
        'D': ['GAC', 'GAT'],
        'E': ['GAG', 'GAA'],
        'F': ['TTT', 'TTC'],
        'G': ['GGT', 'GGG', 'GGA', 'GGC'],
        'H': ['CAT', 'CAC'],
        'I': ['ATC', 'ATA', 'ATT'],
        'K': ['AAG', 'AAA'],
        'L': ['CTT', 'CTG', 'CTA', 'CTC', 'TTA', 'TTG'],
        'M': ['ATG'],
        'N': ['AAC', 'AAT'],
        'P': ['CCT', 'CCG', 'CCA', 'CCC'],
        'Q': ['CAA', 'CAG'],
        'R': ['AGG', 'AGA', 'CGA', 'CGC', 'CGG', 'CGT'],
        'S': ['AGC', 'AGT', 'TCT', 'TCG', 'TCC', 'TCA'],
        'T': ['ACA', 'ACG', 'ACT', 'ACC'],
        'V': ['GTA', 'GTC', 'GTG', 'GTT'],
        'W': ['TGG'],
        'Y': ['TAT', 'TAC'],
        '*': ['TAA', 'TGA', 'TAG']
    }


def back_translate(protein_sequence):
    return ''.join([back_translation_code[aa][0] for aa in protein_sequence])
```

```python

seq_records = []
for index, row in final_sample.iterrows():
    dna_sequence = back_translate(row['Sequence'])
    dseq_obj = Dseq(dna_sequence)
    dseqrecord = Dseqrecord(dseq_obj, name=row['SecretoGen_ID'], description=f"Perplexity: {row['Perplexity']}, Length: {row['Length']}")
    seq_records.append(dseqrecord)

seq_records
```

### Codon optimize the sequences: 

A. oryzae tax id : 5062

https://www.ncbi.nlm.nih.gov/Taxonomy/Browser/wwwtax.cgi?mode=Info&id=5062&lvl=3&lin=f&keep=1&srchmode=1&unlock

```python
# to make the codon optimization reproducible we can initialize a random seed
import numpy as np
np.random.seed(2)
```

```python
from dnachisel import DnaOptimizationProblem, CodonOptimize, EnforceTranslation, EnforceGCContent

# Codon optimization for Aspergillus oryzae
codon_optimized_records = []
codon_optimized_records_str = []
taxonomy_id = 5062 # a oryzae

for seq in seq_records:
    problem = DnaOptimizationProblem(
        sequence=str(seq.seq),
        constraints=[
            EnforceGCContent(mini=0.3, maxi=0.7, window=len(seq)),
            EnforceTranslation(location=(0, len(seq)))
        ],
        #objectives=[CodonOptimize(codon_usage_table=a_oryzae, method='match_codon_usage')], 
        objectives=[CodonOptimize(species=taxonomy_id, method='match_codon_usage')], 
    )
    problem.resolve_constraints()
    
    problem.optimize()
    # PRINT SUMMARIES TO CHECK THAT CONSTRAINTS PASS
    print(problem.constraints_text_summary())
    print(problem.objectives_text_summary())

    # Save the sequences
    optimized_dna_sequence = str(problem.sequence)
    
    dseq_obj = Dseq(optimized_dna_sequence)
    dseqrecord = Dseqrecord(dseq_obj, name=row['SecretoGen_ID'], description=f"Perplexity: {row['Perplexity']}, Length: {row['Length']}")
    codon_optimized_records.append(dseqrecord)
    codon_optimized_records_str.append(str(dseqrecord.seq))
    
final_sample['codon_optimized_sequences']= codon_optimized_records_str
```

```python
# Function to find differences in a specific column between two DataFrames
def find_column_differences(df1, df2, column_name):
    # Check if the specified column exists in both DataFrames
    if column_name not in df1.columns or column_name not in df2.columns:
        print(f"Column '{column_name}' does not exist in both DataFrames.")
        return
    
    # Check if both DataFrames have the same number of rows
    if len(df1) != len(df2):
        print("DataFrames do not have the same number of rows. Exiting comparison.")
        return
    
    # Create a DataFrame to store differences
    diff_df = pd.DataFrame(index=df1.index)
    
    # Find where values differ in the specified column
    mask = df1[column_name] != df2[column_name]
    
    # Store differences in the diff_df
    diff_df[column_name] = df1[column_name].where(mask, other=f"{df1[column_name]} (same)")
    diff_df['Difference'] = df2[column_name].where(mask, other='(same)')
    
    # Remove rows with no differences (all values are marked as 'same')
    diff_df = diff_df[mask]
    
    return diff_df

# Finding and displaying differences in the 'codon_optimized_sequences' column
column_name = 'codon_optimized_sequences'
differences = find_column_differences(final_sample, old_df, column_name)
print(f"Differences in column '{column_name}':")
print(differences)
```

```python

def find_duplicates_between_dfs(df1, df2, columns):
    """
    Find duplicate rows between two DataFrames based on specified columns.
    
    Parameters:
    - df1 (pd.DataFrame): The first DataFrame (e.g., final_sample).
    - df2 (pd.DataFrame): The second DataFrame (e.g., selected_rows).
    - columns (list): List of column names to consider for duplication.
    
    Returns:
    - duplicates_df (pd.DataFrame): Rows from df2 that are duplicates in df1.
    """
    # Ensure the specified columns exist in both DataFrames
    missing_cols_df1 = set(columns) - set(df1.columns)
    missing_cols_df2 = set(columns) - set(df2.columns)
    
    if missing_cols_df1:
        raise ValueError(f"The following columns are missing in df1: {missing_cols_df1}")
    if missing_cols_df2:
        raise ValueError(f"The following columns are missing in df2: {missing_cols_df2}")
    
    # Create a unique identifier by concatenating the specified columns
    df1_unique = df1[columns].drop_duplicates()
    df2_unique = df2[columns].drop_duplicates()
    
    # Merge df2 with df1 to find duplicates
    duplicates_df = pd.merge(df2_unique, df1_unique, on=columns, how='inner', indicator=False)
    
    # To get the full rows from df2 that are duplicates in df1
    # Merge with the original df2 to preserve all columns
    duplicates_full = pd.merge(df2, duplicates_df, on=columns, how='inner')
    
    return duplicates_full

duplicate_columns = ['Sequence', 'SecretoGen_ID']

duplicates = find_duplicates_between_dfs(final_sample, selected_rows, duplicate_columns)
duplicates
```

```python

def report_duplicates(duplicates_df):
    """
    Print a summary and the duplicate rows.
    
    Parameters:
    - duplicates_df (pd.DataFrame): DataFrame containing duplicate rows.
    """
    if duplicates_df.empty:
        print("No duplicates found between the two DataFrames based on the specified columns.")
    else:
        print(f"Found {len(duplicates_df)} duplicate(s) between the DataFrames based on the specified columns.")
        print("Duplicate Rows:")
        print(duplicates_df)
    
report_duplicates(duplicates)
```
