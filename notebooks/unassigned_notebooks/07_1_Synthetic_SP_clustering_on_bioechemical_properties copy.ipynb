# UMAP calculation

```python
import pandas as pd
import umap.umap_ as umap
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import hdbscan
import seaborn as sns
import matplotlib.colors as mcolors


# Prepare the DataFrame by dropping non-numeric columns
data_for_umap = df.drop(columns=['SP_ID', ])

# Standardize the features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data_for_umap)

# Apply UMAP
umap_model = umap.UMAP(n_neighbors=15, min_dist=0.1, n_components=2, random_state=12345)
umap_results = umap_model.fit_transform(scaled_data)

# Create a DataFrame from UMAP results
umap_df = pd.DataFrame(umap_results, columns=['UMAP1', 'UMAP2'])

# Concatenate the UMAP results with the original DataFrame
df_with_umap = pd.concat([df.reset_index(drop=True), umap_df], axis=1)
output_csv_path = '../data/02_all_signal_peptides/biochbiochemical_results_synthetic_signalpeptides_with_UMAP.csv'
df_with_umap.to_csv(output_csv_path, index=False)
```

## PCA and t-SNE

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans

# Assuming 'df' is your DataFrame

# Prepare the DataFrame by dropping the non-numeric column
data_for_analysis = df.drop(columns=['SP_ID'])

# Standardize the features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data_for_analysis)

# Apply PCA
pca = PCA(n_components=2)
pca_results = pca.fit_transform(scaled_data)

# Apply t-SNE
tsne = TSNE(n_components=2, perplexity=30, learning_rate=200, random_state=1234)
tsne_results = tsne.fit_transform(scaled_data)

# Apply K-means clustering
n_clusters = 3  # Adjust based on your data
kmeans = KMeans(n_clusters=n_clusters, random_state=1234)
pca_clusters = kmeans.fit_predict(pca_results)
tsne_clusters = kmeans.fit_predict(tsne_results)

# Set style for plots
sns.set(style='white', context='notebook')

# Plot PCA results
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.scatter(pca_results[:, 0], pca_results[:, 1], c=pca_clusters, cmap='viridis', s=50, alpha=0.6, edgecolor='k')
plt.title('PCA with K-means Clustering', fontsize=14, fontweight='bold')
plt.xlabel('PCA Component 1', fontsize=12)
plt.ylabel('PCA Component 2', fontsize=12)

# Plot t-SNE results
plt.subplot(1, 2, 2)
plt.scatter(tsne_results[:, 0], tsne_results[:, 1], c=tsne_clusters, cmap='viridis', s=50, alpha=0.6, edgecolor='k')
plt.title('t-SNE with K-means Clustering', fontsize=14, fontweight='bold')
plt.xlabel('t-SNE Component 1', fontsize=12)
plt.ylabel('t-SNE Component 2', fontsize=12)

# Adjust layout
plt.tight_layout()

# Save the figure (optional)
plt.savefig('../data/plots/PCA_tSNE_Clustering.png', dpi=300, bbox_inches='tight')

plt.show()
```

## Adjusted Rand Index (ARI) or Normalized Mutual Information (NMI)

```python
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score

# Calculate Adjusted Rand Index
ari = adjusted_rand_score(df['tsne_cluster'], df['umap_cluster'])

# Calculate Normalized Mutual Information
nmi = normalized_mutual_info_score(df['tsne_cluster'], df['umap_cluster'])

print(f"Adjusted Rand Index: {ari:.3f}")
print(f"Normalized Mutual Information: {nmi:.3f}")
```

## Filtering out gene ids


### Understanding ARI and NMI:

- **Adjusted Rand Index (ARI)**: Measures the similarity of the two assignments, ignoring permutations and with chance normalization. An ARI score of 1 means perfect labeling alignment, while 0 or negative values indicate no better than random chance.

- **Normalized Mutual Information (NMI)**: Also measures the similarity, but it's normalized to account for chance. Like ARI, a score of 1 indicates perfect agreement, and lower values indicate less agreement.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans
from scipy.spatial.distance import euclidean

# Define the list of sampled natural signal peptides
martis_list = [
    "AO090003001591-T-p1",
    "AO090005001090-T-p1",
    "AO090023000905-T-p1",
    "AO090038000279-T-p1",
    "AO090701000295-T-p1",
    "AO090003001036-T-p1",
    "AO090001000635-T-p1",
    "AO090005000409-T-p1",
    "AO090009000117-T-p1",
    "AO090102000440-T-p1"
]

# Load the natural and synthetic sps dataframes
natural_sps_df = pd.read_csv('../data/02_all_signal_peptides/biochemical_result_w_abundance.csv', index_col=False)
synthetic_sps_df = pd.read_csv('../data/02_all_signal_peptides/biochemical_results_synthetic_signalpeptides.csv', index_col=False)
chosen_peptides_df = pd.read_excel('../data/09_primers/full_dataframe_of_sampled_SPs_2024_08_02.xlsx')
chosen_peptides_ids = chosen_peptides_df['SecretoGen_ID'].tolist()

# Load the chosen signal peptides SECOND ROUND
chosen_peptides_df_2 = pd.read_excel('../data/09_primers/full_dataframe_of_sampled_SPs_2024_14_11.xlsx')
chosen_peptides_ids_2 = chosen_peptides_df_2['SecretoGen_ID'].tolist()

# Concatenate the two DataFrames
chosen_peptides_df = pd.concat([chosen_peptides_df, chosen_peptides_df_2], ignore_index=True)
chosen_peptides_ids = chosen_peptides_ids + chosen_peptides_ids_2

# Mark the chosen peptides in the synthetic dataframe
synthetic_sps_df['Type'] = 'Synthetic'
synthetic_sps_df.loc[synthetic_sps_df['SP_ID'].isin(chosen_peptides_ids), 'Type'] = 'Sampled_Synthetic'

# Mark the sampled natural signal peptides
natural_sps_df['Type'] = 'Natural'
natural_sps_df.loc[natural_sps_df['gene'].isin(martis_list), 'Type'] = 'Sampled_Natural'

# Combine the natural, synthetic, and chosen data for scaling
combined_df = pd.concat([
    natural_sps_df.drop(columns=['gene', 'abundance']),
    synthetic_sps_df.drop(columns=['SP_ID'])
])

# Standardize the features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(combined_df.drop(columns=['Type']))

# Apply t-SNE with the same random state and settings as your original plot
tsne = TSNE(n_components=2, perplexity=30, learning_rate=200, random_state=123)
tsne_results = tsne.fit_transform(scaled_data)

# Create a combined dataframe for plotting
tsne_df = pd.DataFrame(tsne_results, columns=['t-SNE Component 1', 't-SNE Component 2'])
tsne_df['Type'] = combined_df['Type'].values

# Apply K-means clustering
n_clusters = 3  # Adjust based on your data
kmeans = KMeans(n_clusters=n_clusters, random_state=123)
tsne_df['Cluster'] = kmeans.fit_predict(tsne_results)

# Calculate the centroids of the natural and synthetic signal peptides
natural_centroid = tsne_df[tsne_df['Type'] == 'Natural'][['t-SNE Component 1', 't-SNE Component 2']].mean().values
synthetic_centroid = tsne_df[tsne_df['Type'] == 'Synthetic'][['t-SNE Component 1', 't-SNE Component 2']].mean().values

# Calculate the Euclidean distance between the centroids
distance = euclidean(natural_centroid, synthetic_centroid)

print(f"The Euclidean distance between the centroids of natural and synthetic signal peptides is: {distance:.2f}")

# Set style for plots
sns.set(style='white', context='notebook')

# Define a publication-worthy color palette for type with distinct colors
palette_type = {
    'Natural': '#4C72B0',            # Blue
    'Synthetic': '#DD8452',          # Orange
    'Sampled_Synthetic': '#55A868',  # Green
    'Sampled_Natural': '#C44E52'     # Red
}

# Define a color palette for clusters
palette_cluster = sns.color_palette("viridis", n_colors=n_clusters)  # 'viridis' is a perceptually uniform color map

# Create the plots
fig, axes = plt.subplots(1, 2, figsize=(20, 10))

# Plot combined t-SNE results by Type
sns.scatterplot(data=tsne_df, x='t-SNE Component 1', y='t-SNE Component 2', hue='Type', style='Type', palette=palette_type, markers=['o', 's', 'D', '^'], s=100, alpha=0.7, edgecolor='k', ax=axes[0])
axes[0].set_title('t-SNE for Natural, Synthetic, Sampled Synthetic, and Sampled Natural Signal Peptides', fontsize=16, fontweight='bold')
axes[0].set_xlabel('t-SNE Component 1', fontsize=14)
axes[0].set_ylabel('t-SNE Component 2', fontsize=14)

# Plot t-SNE results by Cluster
sns.scatterplot(data=tsne_df, x='t-SNE Component 1', y='t-SNE Component 2', hue='Cluster', palette=palette_cluster, s=100, alpha=0.7, edgecolor='k', ax=axes[1])
axes[1].set_title('t-SNE with K-means Clustering', fontsize=16, fontweight='bold')
axes[1].set_xlabel('t-SNE Component 1', fontsize=14)
axes[1].set_ylabel('t-SNE Component 2', fontsize=14)

# Adjust layout
plt.tight_layout()

# Save the figure (optional)
plt.savefig('../data/plots/natural_vs_synthetic_vs_sampled_synthetic_vs_sampled_natural_tSNE_Clustering_combined_with_clusters.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.spatial.distance import euclidean
import umap  # Import UMAP

# Define the list of sampled natural signal peptides
martis_list = [
    "AO090003001591-T-p1",
    "AO090005001090-T-p1",
    "AO090023000905-T-p1",
    "AO090038000279-T-p1",
    "AO090701000295-T-p1",
    "AO090003001036-T-p1",
    "AO090001000635-T-p1",
    "AO090005000409-T-p1",
    "AO090009000117-T-p1",
    "AO090102000440-T-p1"
]

# Load the natural and synthetic sps dataframes
natural_sps_df = pd.read_csv('../data/02_all_signal_peptides/biochemical_result_w_abundance.csv', index_col=False)
synthetic_sps_df = pd.read_csv('../data/02_all_signal_peptides/biochemical_results_synthetic_signalpeptides.csv', index_col=False)
chosen_peptides_df = pd.read_excel('../data/09_primers/full_dataframe_of_sampled_SPs_2024_08_02.xlsx')
chosen_peptides_ids = chosen_peptides_df['SecretoGen_ID'].tolist()

# Load the chosen signal peptides SECOND ROUND
chosen_peptides_df_2 = pd.read_excel('../data/09_primers/full_dataframe_of_sampled_SPs_2024_14_11.xlsx')
chosen_peptides_ids_2 = chosen_peptides_df_2['SecretoGen_ID'].tolist()

# Concatenate the two DataFrames
chosen_peptides_df = pd.concat([chosen_peptides_df, chosen_peptides_df_2], ignore_index=True)
chosen_peptides_ids = chosen_peptides_ids + chosen_peptides_ids_2

# Mark the chosen peptides in the synthetic dataframe
synthetic_sps_df['Type'] = 'Synthetic'
synthetic_sps_df.loc[synthetic_sps_df['SP_ID'].isin(chosen_peptides_ids), 'Type'] = 'Sampled_Synthetic'

# Mark the sampled natural signal peptides
natural_sps_df['Type'] = 'Natural'
natural_sps_df.loc[natural_sps_df['gene'].isin(martis_list), 'Type'] = 'Sampled_Natural'

# Combine the natural, synthetic, and chosen data for scaling
combined_df = pd.concat([
    natural_sps_df.drop(columns=['gene', 'abundance']),
    synthetic_sps_df.drop(columns=['SP_ID'],)])

# Standardize the features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(combined_df.drop(columns=['Type']))

# Initialize UMAP with desired parameters
umap_model = umap.UMAP(
    n_components=2,        # Number of dimensions for the projection
    n_neighbors=15,        # Controls the local neighborhood size
    min_dist=0.1,          # Controls the minimum distance between points
    metric='euclidean',    # Distance metric
    random_state=12345     # Ensures reproducibility
)

# Apply UMAP to the scaled data
umap_results = umap_model.fit_transform(scaled_data)

# Create a combined dataframe for plotting
umap_df = pd.DataFrame(umap_results, columns=['UMAP Component 1', 'UMAP Component 2'])
umap_df['Type'] = combined_df['Type'].values

# Save UMAP results for consistent use across multiple codes
umap_df.to_csv('../data/umap_results.csv', index=False)

# Apply K-means clustering (optional)
n_clusters = 3  # Adjust based on your data
kmeans = KMeans(n_clusters=n_clusters, random_state=123)
umap_df['Cluster'] = kmeans.fit_predict(umap_results)

# Calculate the centroids of the natural and synthetic signal peptides
natural_centroid = umap_df[umap_df['Type'] == 'Natural'][['UMAP Component 1', 'UMAP Component 2']].mean().values
synthetic_centroid = umap_df[umap_df['Type'] == 'Synthetic'][['UMAP Component 1', 'UMAP Component 2']].mean().values

# Calculate the Euclidean distance between the centroids
distance = euclidean(natural_centroid, synthetic_centroid)

print(f"The Euclidean distance between the centroids of natural and synthetic signal peptides is: {distance:.2f}")

# Set style for plots
sns.set(style='white', context='notebook')

# Define a publication-worthy color palette for type with distinct colors
palette_type = {
    'Natural': '#4C72B0',            # Blue
    'Synthetic': '#DD8452',          # Orange
    'Sampled_Synthetic': '#55A868',  # Green
    'Sampled_Natural': '#C44E52'     # Red
}

# Define a color palette for clusters
palette_cluster = sns.color_palette("viridis", n_colors=n_clusters)  # 'viridis' is a perceptually uniform color map

# Create the plots
fig, axes = plt.subplots(1, 2, figsize=(20, 10))

# Plot combined UMAP results by Type
sns.scatterplot(
    data=umap_df, 
    x='UMAP Component 1', 
    y='UMAP Component 2', 
    hue='Type', 
    style='Type', 
    palette=palette_type, 
    markers=['o', 's', 'D', '^'], 
    s=100, 
    alpha=0.7, 
    edgecolor='k', 
    ax=axes[0]
)
axes[0].set_title('UMAP for Natural, Synthetic, Sampled Synthetic, and Sampled Natural Signal Peptides', fontsize=16, fontweight='bold')
axes[0].set_xlabel('UMAP Component 1', fontsize=14)
axes[0].set_ylabel('UMAP Component 2', fontsize=14)

# Plot UMAP results by Cluster
sns.scatterplot(
    data=umap_df, 
    x='UMAP Component 1', 
    y='UMAP Component 2', 
    hue='Cluster', 
    palette=palette_cluster, 
    s=100, 
    alpha=0.7, 
    edgecolor='k', 
    ax=axes[1]
)
axes[1].set_title('UMAP with K-means Clustering', fontsize=16, fontweight='bold')
axes[1].set_xlabel('UMAP Component 1', fontsize=14)
axes[1].set_ylabel('UMAP Component 2', fontsize=14)

# Adjust layout
plt.tight_layout()

# Save the figure (optional)
plt.savefig('../data/plots/natural_vs_synthetic_vs_sampled_synthetic_vs_sampled_natural_UMAP_Clustering_combined_with_clusters.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_ind

# Define the palette for consistent colors
palette_type = {
    'Natural': '#4C72B0',            # Blue
    'Synthetic': '#DD8452',          # Orange
    'Sampled_Synthetic': '#55A868',  # Green
    'Sampled_Natural': '#C44E52'     # Red
}

# Calculate the sample sizes
natural_count = combined_df[combined_df['Type'] == 'Natural'].shape[0]
synthetic_count = combined_df[combined_df['Type'] == 'Synthetic'].shape[0]
sampled_synthetic_count = combined_df[combined_df['Type'] == 'Sampled_Synthetic'].shape[0]
sampled_natural_count = combined_df[combined_df['Type'] == 'Sampled_Natural'].shape[0]

# Perform statistical tests to compare distributions and mark significant differences
results = []
features = combined_df.drop(columns=['Type']).columns

for feature in features:
    natural_values = combined_df[combined_df['Type'] == 'Natural'][feature]
    synthetic_values = combined_df[combined_df['Type'] == 'Synthetic'][feature]
    t_stat, p_value = ttest_ind(natural_values, synthetic_values)
    significance = '*' if p_value < 0.05 else ''
    results.append({'Feature': feature, 't-statistic': t_stat, 'p-value': p_value, 'Significance': significance})

results_df = pd.DataFrame(results)
print("Statistical test results comparing natural and synthetic signal peptides:")
print(results_df)
results_df.to_csv('../data/plots/statistical_test_results.csv', index=False)

# Create a main title with sample sizes
main_title = f'Distribution of Features by Type\n(Natural: n={natural_count}, Synthetic: n={synthetic_count}, Sampled_Synthetic: n={sampled_synthetic_count}, Sampled_Natural: n={sampled_natural_count})'

# Compare distributions of features using box plots with consistent colors and significance annotations
plt.figure(figsize=(20, 15))
plt.suptitle(main_title, fontsize=16, fontweight='bold', y=1.02)

for i, feature in enumerate(features, 1):
    plt.subplot(4, 3, i)
    ax = sns.boxplot(x='Type', y=feature, data=combined_df, hue='Type', palette=palette_type, legend=False)
    plt.title(f'{feature}', fontsize=12)
    
    # Add significance annotation
    if results_df.loc[results_df['Feature'] == feature, 'Significance'].values[0] == '*':
        y, h, col = combined_df[feature].max() * 1.05, combined_df[feature].max() * 0.03, 'k'
        plt.plot([0, 1], [y, y], lw=1.5, c=col)
        plt.text(0.5, y + h, '*', ha='center', va='bottom', color=col)

plt.tight_layout(rect=[0, 0, 1, 0.95])  # Adjust layout to make room for the main title
plt.savefig('../data/plots/feature_distributions_by_type_with_significance.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Function to get localization counts
def get_localization_counts(df):
    return df['Localizations'].value_counts()

# Get counts for each cluster
counts1 = get_localization_counts(cluster1)
counts2 = get_localization_counts(cluster2)
counts3 = get_localization_counts(cluster3)

# Set the aesthetics for the plots
sns.set(style="whitegrid")
plt.rcParams.update({'font.size': 10, 'font.family': 'sans-serif'})

# Create subplots with larger figure size
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(25, 8), sharey=True)

# Define a color palette with enough colors
max_unique_values = max(counts1.size, counts2.size, counts3.size)
palette = sns.color_palette("husl", max_unique_values)

# Function to create a barplot for a cluster
def create_barplot(ax, counts, title):
    sns.barplot(ax=ax, x=counts.index, y=counts.values, palette=palette[:counts.size])
    ax.set_title(title, fontsize=14, fontweight='bold')
    ax.set_xlabel('Localizations', fontsize=12)
    ax.set_ylabel('Count', fontsize=12)
    ax.tick_params(axis='x', rotation=90)  # Rotate labels

# Create bar plots for each cluster
create_barplot(axes[0], counts1, 'Cluster 1 Localizations')
create_barplot(axes[1], counts2, 'Cluster 2 Localizations')
create_barplot(axes[2], counts3, 'Cluster 3 Localizations')

# Adjust layout and spacing
plt.tight_layout()
plt.subplots_adjust(wspace=0.1)  # Adjust horizontal spacing
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Assuming df1, df2, df3 are your dataframes

def plot_top_bgd_genes(df, title, top_n=10):
    # Sort the dataframe by 'Result count' and select top N
    df_sorted = df.sort_values(by='Result count', ascending=False).head(top_n)

    # Setting the aesthetic style for the plots
    sns.set(style='whitegrid', context='talk', palette='viridis')

    # Plotting
    plt.figure(figsize=(8, 4))
    sns.barplot(x='Result count', y='Name', data=df_sorted, edgecolor='black')
    plt.title(f'Top {top_n} Used Genes in Background for {title}', fontsize=16, fontweight='bold')
    plt.xlabel('Background Gene Count', fontsize=14)
    plt.ylabel('GO Term', fontsize=14)
    plt.xticks(fontsize=12)
    plt.yticks(fontsize=12)
    sns.despine()  # Removes top and right spines for a cleaner plot

    # Save the figure (optional)
    plt.savefig(f'../data/plots/{title}_Top_Background_Genes.png', dpi=300, bbox_inches='tight')

    plt.show()

# Plotting for each DataFrame
plot_top_bgd_genes(df1, 'Group1')
plot_top_bgd_genes(df2, 'Group2')
plot_top_bgd_genes(df3, 'Group3')
```
